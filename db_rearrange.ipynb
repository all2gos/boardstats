```python
import pandas as pd
import os

pwd = os.getcwd()
files = [x for x in os.listdir(pwd) if '.csv' in x]

df = pd.DataFrame()

for file in files:
        with open(pwd + '/' + file,'r',encoding='utf-8') as f:
            data = pd.read_csv(f)            
            df = pd.concat([df,data])
df = df.drop('Unnamed: 0',axis=1).drop_duplicates()
df = df.reset_index(drop=True)



df.to_csv('board_df.csv')
print('plik po synchronizacji zawiera',len(df),'wpisów')

```
